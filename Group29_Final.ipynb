## Dataset Description and Loading

```python
# --- Data Extraction Function ---
# Here we define a function to extract the chosen relevant data features from raw CSV files 1 for each season.
# This function will look for the average odds columns and handle cases where they might be missing.

def extract_data(df, filename):
    """
    Extracts premier league data from CSV file.
    Prefers BbAvH/D/A, otherwise falls back to AvgH/D/A.
    Adds season column extracted from filename.

    parameters:
    df: DataFrame containing the raw data from the CSV file.
    filename: The name of the CSV file, used to extract the season information.

    return: A DataFrame with the extracted features and season information, or None if required columns are missing.

    """

    primary_cols = ['HomeTeam', 'AwayTeam', 'FTR', 'Referee', 'FTHG', 
                  'FTAG', 'HS', 'AS', 'HST', 'AST', 'HF', 'AF',
                'HC', 'AC', 'HY', 'AY', 'HR', 'AR', "BbAvH", "BbAvD", "BbAvA", 'Date']
    fallback_cols = ['HomeTeam', 'AwayTeam', 'FTR', 'Referee', 'FTHG', 
                  'FTAG', 'HS', 'AS', 'HST', 'AST', 'HF', 'AF',
                'HC', 'AC', 'HY', 'AY', 'HR', 'AR', "AvgH", "AvgD", "AvgA", 'Date']

    season = os.path.splitext(filename)[0]

    # Trying to find Betbrain columns first
    if all(col in df.columns for col in primary_cols):
        result = df[primary_cols].copy()
        result.columns = ['HomeTeam', 'AwayTeam', 'FTR', 'Referee', 'FTHG', 
                            'FTAG', 'HS', 'AS', 'HST', 'AST', 'HF', 'AF',
                            'HC', 'AC', 'HY', 'AY', 'HR', 'AR', "AvgH", "AvgD", "AvgA", 'Date']
        print(f'Found primary columns: BbAvH, BbAvD, BbAvA for {filename}')

    # Else trying to find fallback columns
    elif all(col in df.columns for col in fallback_cols):
        result = df[fallback_cols].copy()
        print(f'Using fallback columns: AvgH, AvgD, AvgA for {filename}')

    # If neither set of columns is found, print a warning and return None
    else:
        print(f"⚠ Warning: Required odds columns not found in {filename}")
        return None

    # Add season column
    result["Season"] = season
    
    return result
```

```python
# --- User settings begin ---
dataPath = "Data" # Define the folder where the raw CSV files are located.
# --- User settings end ---

new_data = []

for file in os.listdir(dataPath):
    if file.endswith(".csv"):
        path = os.path.join(dataPath, file)
        try:
            df = pd.read_csv(path)
        except Exception as e:
            print(f"Error reading {file}: {e}")
            continue

        extracted = extract_data(df, file)

        if extracted is not None:
            new_data.append(extracted)

# Combine everything into one dataframe
final_df = pd.concat(new_data, ignore_index=True)
```

```python
print(f"\n Final dataset:")
final_df = final_df.dropna(subset=["HomeTeam", "AwayTeam"]) # 2014/2015 season has a row of NaN that is removed here
final_df
```

## Ground truth encoding

```python
def encode_results(results):
    """
    Endcoding of results
    1: HOME WIN, -1: AWAY WIN, 0: DRAW

    Parameters:
        results (pandas Series): results to encode

    Returns:
        encoding (numpy array): encoded results
    """
    encoding = np.zeros(len(results)) # Initialize encoding array

    for i in range(len(results)):
        if results[i] == "H":
            encoding[i] = 1 # Home win
        elif results[i] == "A":
            encoding[i] = -1 # Away win
        else:
            encoding[i] = 0 # Draw
    return encoding
data = final_df.copy()
ground_truth = encode_results(data['FTR'])
print("Amount of home wins: ", np.sum(ground_truth == 1))
print("Amount of away wins: ", np.sum(ground_truth == -1))
print("Amount of draws: ", np.sum(ground_truth == 0))
```

## DBSCAN

### Tuning the parameter $\epsilon$
Before running DBSCAN on the dataset, we try to find the best hyperparameter for $\epsilon$.

First we want to find the "border object" or "first valley" (non differentiable point). This is done by defining a function that for each point find the distance to the k-nearest neighbor. This distance is the distance to the MinPts-nearest nighbor where MinPts is set to $2d-1$ where $d$ is the number of features (Slide 22 lecture 3).

The distance measure used, is both the euclediean distance and manhattan distance.

Afterwards we plot the distances in descending order.

```python
# -- Calculating K nearest neighbours --
def find_k_nearest_neighbor_dist(data, dist: str = 'euclidean', minpts=None, use_tqdm=True):
    """
    Compute the distance to the k-th nearest neighbor for each point.

    Parameters:
        data (numpy array): Shape (n_samples, d_features)
        dist (str): The distance metric to use (default 'euclidean')
        minpts (int): Number of neighbors to consider (default 2*d - 1 -- See slide 22 lecture 3)
        use_tqdm (bool): Show progress bar for large datasets
        minpts (int): Number of neighbors to consider (default 2*d - 1 -- See slide 22 lecture 3)
        use_tqdm (bool): Show progress bar for large datasets

    Returns:
        distances_sorted (numpy array): k-th nearest neighbor distances sorted descending
        minpts (int): The k used
    """
    n, d = data.shape
    if minpts is None:
        minpts = 2 * d - 1

    # Compute full pairwise distance matrix (n x n)
    pairwise_dist = cdist(data, data, metric=dist) # Manhattan distance is used

    np.fill_diagonal(pairwise_dist, np.inf) # Set diagonal to np.inf to ignore self-distance

    distances = np.zeros(n)

    iterator = tqdm(range(n), disable=not use_tqdm)
    for i in iterator:
        # Sort distances for this point
        sorted_dists = np.partition(pairwise_dist[i], minpts)  # faster than full sort
        distances[i] = sorted_dists[minpts]  # minpts-th nearest neighbor

    # Return distances sorted descending
    return np.sort(distances)[::-1], minpts



y_man, minpts_man = find_k_nearest_neighbor_dist(modified_stats.values, dist='cityblock')
print('Found value for minpts (Manhattan):', minpts_man)

y_euclidean, minpts_euclidean = find_k_nearest_neighbor_dist(modified_stats.values, dist='euclidean')
print('Found value for minpts (Euclidean):', minpts_euclidean)
```

```python
# -- Plotting K nearest Neighbor distances---
x_man = np.arange(len(y_man))
x_euclidean = np.arange(len(y_euclidean))
epsilon_man = 2.7
epsilon_euclidean = 1.8
fig, ax = plt.subplots(1,2, figsize=(10, 5))
ax[0].plot(x_man, y_man)
ax[0].set_xlabel("Point Index")
ax[0].set_ylabel(f"Distance to {minpts_man}th Nearest Neighbor")
ax[0].set_title(f"Manhattan distance")
ax[0].axhline(epsilon_man, c='k', ls='--', label=f'Epsilon = {epsilon_man}')
ax[0].legend()

ax[1].plot(x_euclidean, y_euclidean)
ax[1].set_xlabel("Point Index")
ax[1].set_ylabel(f"Distance to {minpts_euclidean}th Nearest Neighbor")
ax[1].set_title(f"Euclidean distance")
ax[1].axhline(epsilon_euclidean, c='k', ls='--', label=f'Epsilon = {epsilon_euclidean}')
ax[1].legend()
plt.tight_layout()
```

As seen above, it is hard to to determine the right $\epsilon$ as there is no clear valley in either of the plots. This is a sign that there isn't clearly defined clusters in the data, as the density acts continously rather than discretely.
However we choose $\epsilon = 1.8$ for the euclidean distance and $\epsilon = 2.7$ for the manhattan distance, and continue to DBSCAN.

### DBSCAN clustering

```python
dbscan = DBSCAN(eps=epsilon_euclidean, min_samples=minpts_euclidean)
dbscan.fit(modified_stats.values)
labels_dbscan = dbscan.labels_
print("DBSCAN Cluster Labels:", np.unique(labels_dbscan)) # -1 corresponds to noise points, other integers correspond to cluster labels
```

It is observed that the result of DBSCAN when using the manhattan distance is exstremely sensitive to the choice of epsilon. Choosing:

This is a sign that the data is ill suited for clustering using the manhattan distance. When using the euclidean distance, the result is more stable. Therefor we choose to continue with the euclidean distance for the rest of the analysis.

There is no clear valley in the plot of the k-nearest neighbor distances, so the clusters obtained may not be well defined. 
As seen above we obtain three clusters using the euclidean distance (0, 1, 2) and some noise (-1).

To evaluate the clustering, we define two functions, one makes the contingency table based on the FTR ground truth, and the other evaluates the purity score.
- The contingency table tells us how many points of each ground truth the clusters contain.
- The purity measure tells us how many points in the cluster belong to the same (most prominent) ground truth.

```python
def contingency_table(C, T, ignore_noise=True): 
    """
    Parameters:
        C(numpy array):       Clusters obtained by a clustering algorithm as a nx1 vector
        T(numpy array):       Ground-truth cluster assignments as a nx1 vector

    Returns:
        ctable (numpy array):   a num_clusters_of_C x num_cluster_of_T matrix containing the overlaps among the different clusters
    """
    
    if ignore_noise:
        mask = (C != -1)
        C_masked = C[mask]
        indices_masked = np.where(mask)[0]
    else:
        C_masked = C
        indices_masked = np.arange(len(C))

    clusters = np.unique(C_masked)
    classes  = np.unique(T)   # include all classes, not just masked
    
    ctable = np.zeros((len(clusters), len(classes)), dtype=int)

    for i, c in enumerate(clusters):
        for j, t in enumerate(classes):
            # count overlap only for points that are not noise
            count = np.sum((C_masked == c) & (T[indices_masked] == t))
            ctable[i, j] = count
    return ctable

def purity_score(ctable):
    """
    Parameters:
        ctable (numpy array):  Contingency table of shape (num_clusters, num_classes) 
            containing the overlaps among the different clusters
    
    Returns:
        purity (numpy array): a float for each cluster, the purity of clustering result
    """
    
    clusters, _ = ctable.shape
    purity = np.zeros(clusters)

    for i in range(clusters):
        purity[i] = np.max(ctable[i])/np.sum(ctable[i]) if np.sum(ctable[i]) > 0 else 0
    return purity



# --- Contingency table to compare the cluster labels with the ground truth ---
c_table = contingency_table(labels_dbscan, ground_truth)
print("Contingency Table:\n", c_table) 

# --- Calculate the purity of the clustering ---
purity = purity_score(c_table)
print(f"Purity score of the clustering: {purity}")
```

First, looking at the contingency table, the rows correspond to different clusters, while the collumns correspond to ground truth (1: Homewin, 2: draw, 3: Homeloose). Since each row in the contingency table is heavily mixed, the clustering seems to not obtain the ground truth. This can also be seen in the purity, which calculates the proportion of elements in the cluster that belong to the most frequent class. All purities is between 45 and 64 percent, meaning no cluster is pure compared to the ground truth.

If one cluster should be highlighted it would be cluster 1. It is smaller but dominated by home losses, giving it the highest purity. This suggests DBSCAN found a small region where matches with similar characteristics tend to result in home losses.

## T-SNE


To investigate whether our features could reveal meaningful clusters, we applied **t-distributed Stochastic Neighbor Embedding (T-SNE)**.  

T-SNE is basically a dimensionality reduction algorithm that is useful for visualizing high-dimensional data. It works by modeling pairwise similarities between points in the high-dimensional space and trying to preserve these similarities in a lower-dimensional space. Essentially, points that are close in the original feature space should remain close in the lower dimension visualization, while distant points stay far apart.

The goal of using T-SNE here was to see if our features could form clusters and investigate properties of these clusters. This analysis serves as a useful check.

In the first cell below, T-SNE algorithm is applied to the modified stats.

# Graph-Based analysis (part II)

**Graph-based analysis objective:**

The objective of the graph-based analysis is to examine whether graph mining techniques can determine team strength from in-game statistics across multiple seasons and identify any anomalous patterns at the team or season level. Each match is represented as an edge, weighted by a combination of performance statistics such as shots, corners, fouls, and cards. By applying PageRank to the resulting graph, we aim to derive a ranking that reflects team strength.

As ground truth, we use a normalised league table, allowing us to evaluate how well the graph-based ranking aligns with the true competitive standing of each team. The closer the PageRank ranking is to the normalised points table, the more effectively the graph captures real team strength.

```python
#---- Computing table of points----
def compute_points_table(df) -> list[tuple[str, float]]:
    """
    Computes the points table from a dataframe
    Parameters:
        df (pd.DataFrame): DataFrame containing at least the columns 'HomeTeam', 'AwayTeam', and 'FTR' (Full Time Result)
    Returns:
        sorted_points (list of tuples): (team, points) sorted by points in descending order
    """
    # Get all unique teams
    teams = pd.unique(df[['HomeTeam', 'AwayTeam']].values.ravel())
    
    # Initialize points dictionary
    points = {team: 0 for team in teams}
    
    # Iterate over matches
    for _, row in df.iterrows():
        if row['FTR'] == 'H':
            points[row['HomeTeam']] += 3
        elif row['FTR'] == 'A':
            points[row['AwayTeam']] += 3
        elif row['FTR'] == 'D':
            points[row['HomeTeam']] += 1
            points[row['AwayTeam']] += 1

    # Convert to sorted list (like standings)
    sorted_points = sorted(points.items(), key=lambda x: x[1], reverse=True)
    return sorted_points


#---- Normalizing table of points----
def normalize_points(points_table: list[tuple[str, float]], df: pd.DataFrame) -> list[tuple[str, float]]:
    """
    The normalize_points function computes two different normalized rankings of teams based on their accumulated points. 
    See documentation in markdown below.

    Parameters:
        points_table (list of tuples): list of tuples (team, points), representing total points earned by each team
        df (pd.DataFrame): DataFrame containing match data with at least the columns 'HomeTeam' and 'AwayTeam'

    Returns:
        total_points_normal (list of tuples): list of tuples (team, normalized_score) where each team's score is divided 
                                            by the total points across all teams, sorted in descending order
        
        points_pr_match (list of tuples): list of tuples (team, normalized_score) where each team's points are first 
                                        divided by the number of games played, then normalized across all teams 
                                        to sum to 1, sorted in descending order
    """
    # Points/match normalization
    games_played = (
        df.groupby('HomeTeam').size()
        .add(df.groupby('AwayTeam').size(), fill_value=0)
        .astype(int)
    )
    points_pr_match = [
        (team, points / (games_played[team]))
        for team, points in points_table
    ]
    normalisation_factor = sum(points for _, points in points_pr_match)
    points_pr_match = [(team, score / normalisation_factor) for team, score in points_pr_match] 

    # Total points normalization
    total_points_all_teams = sum(points for _, points in points_table)
    
    
    total_points_normal = [(team, points / total_points_all_teams) for team, points in points_table]
    
    return sorted(total_points_normal, key=lambda x: x[1], reverse=True), sorted(points_pr_match, key=lambda x: x[1], reverse=True)

norm_pts_1, norm_pts_2 = normalize_points(compute_points_table(data), data)
assert abs(sum(score for _, score in norm_pts_1) - 1.0) < 1e-10
assert abs(sum(score for _, score in norm_pts_2) - 1.0) < 1e-10
```

## Baseline method
We start by calculating the pagerank of an undirected unweighted graph where each unique team is nodes and if two teams have played eachother they have an edge. The graph is plotted above.

In the following we have investigated two ways of calculating the loss and two ways of normalizing the points table:
### Loss functions
**Rank Loss**
Compares the ordering produced by PageRank against the league table ordering. For each team, it computes the squared difference between its rank position in the PageRank output and its rank position in the points table, then averages across all teams. A perfect score of 0 means PageRank reproduces the league standings in exactly the correct order.

**Score Loss**
Compares the values produced by PageRank directly against the normalised points. For each team, it computes the squared difference between its raw PageRank score and its normalised points value, then averages across all teams. This penalises not just ordering mistakes but also how far apart the two scores are numerically.

### Ground truth normalisation methods
**points pr. match Normalisation (norm_pts_1)**
Each team's points are first divided by the number of played matches, and later normalized giving a win rate bounded between 0 and 1 and summing to 1 across all teams. This accounts for teams that played different numbers of matches.

**Total points Normalisation (norm_pts_2)**
Each team's points are divided by the total points distributed across all teams in the season, giving each team a share of the overall points pool. A team's score therefore represents its fraction of all points won, and the scores sum to 1 across the league.



```python
def print_ranking_comparison(
    pagerank_ranking: list[tuple[str, float]],
    points_rankings: dict[str, list[tuple[str, float]]],
    losses: dict[str, float] | None = None,
) -> None:
    """
    Side-by-side comparison of PageRank ranking vs. an arbitrary number of reference rankings.

    Parameters:
        pagerank_ranking (list of tuples): list of (team, score) sorted by PageRank descending.
        points_rankings (dict): dict of {label: list of (team, score)}, each sorted descending.
        losses (dict | None): optional dict of {label: loss_value} to print below the table.
    """
    col_width = 25
    score_width = 10

    header = f"{'Rank':<6} {'PageRank Team':<{col_width}} {'PR Score':<12}"
    for label in points_rankings:
        header += f" {label:<{col_width}} {'Score':<{score_width}}"
    print(header)
    print("-" * (6 + 12 + col_width + (col_width + score_width + 2) * len(points_rankings)))

    for i, (pr_team, pr_score) in enumerate(pagerank_ranking, 1):
        row = f"{i:<6} {pr_team:<{col_width}} {pr_score:<12.4f}"
        for ranking in points_rankings.values():
            if i <= len(ranking):
                pt_team, pt_score = ranking[i - 1]
                row += f" {pt_team:<{col_width}} {pt_score:<{score_width}.4f}"
        print(row)

    print("-" * (6 + 12 + col_width + (col_width + score_width + 2) * len(points_rankings)))

    if losses:
        for loss_name, loss_value in losses.items():
            print(f"Loss {loss_name}: {loss_value:.6f}")

def rank_loss(pagerank_scores, norm_pts):
    """
    Loss function that penalises rank disagreement between PageRank and normalised points.
    For each team, the penalty is the squared difference in rank position between the two tables.
    E.g. if Liverpool is rank 1 in PageRank and rank 2 in the points table, the penalty is (1-2)^2 = 1.

    Parameters:
        pagerank_scores (list of tuples): list of tuples (team, pagerank_score) sorted by score descending.
        norm_pts (list of tuples): list of tuples (team, normalized_points) sorted by score descending.

    Returns:
        loss_value (float): mean squared rank difference across all teams present in both tables.
    """
    pr_ranks  = {team: rank for rank, (team, _) in enumerate(pagerank_scores, 1)}
    pts_ranks = {team: rank for rank, (team, _) in enumerate(norm_pts, 1)}

    teams = [t for t in pr_ranks if t in pts_ranks]
    return sum((pr_ranks[t] - pts_ranks[t]) ** 2 for t in teams) / len(teams)

def score_loss(pagerank_scores, norm_pts):
    """
    Loss function to evaluate how well the pagerank scores align with the normalised points.
    Parameters:
        pagerank_scores (list of tuples): list of tuples (team, pagerank_score) sorted by score descending.
        norm_pts (list of tuples): list of tuples (team, normalized_points) sorted by score descending.
    Returns:
        loss_value (float): the mean squared error between pagerank scores and normalized points for teams
    """
    pr_dict  = dict(pagerank_scores)
    pts_dict = dict(norm_pts)
    teams    = [t for t in pr_dict if t in pts_dict]
    return sum((pr_dict[t] - pts_dict[t]) ** 2 for t in teams) / len(teams)        

pR = nx.pagerank(base_G, alpha=0.85)
sorted_pagerank = dict(sorted(pR.items(), key=lambda x: x[1], reverse=True))
base_loss_1 = score_loss(list(sorted_pagerank.items()), norm_pts_1)
base_loss_2 = score_loss(list(sorted_pagerank.items()), norm_pts_2)
base_table_loss_1, base_table_loss_2 = rank_loss(list(sorted_pagerank.items()), norm_pts_1), rank_loss(list(sorted_pagerank.items()), norm_pts_2)
losses = {
    "(score) Total points normalized": base_loss_1,
    "(score) Points/match normalized": base_loss_2,
    "(rank) total points": base_table_loss_1,
    "(rank) points/match": base_table_loss_2
}
ground_truths = {
    "Total points normalized": norm_pts_1,
    "Points/match normalized": norm_pts_2
}

print_ranking_comparison(list(sorted_pagerank.items()), ground_truths, losses)
```

## More sophisticated graph-based method
To capture more nuanced relationships between teams, we construct a weighted directed graph where edges represent matches and weights are derived from in-game performance statistics. For each match, we calculate a composite score based on features such as shot difference, corner difference, foul difference, and card difference. The composite score is essentially an inner product between the feature vector and a set of learned weights (in the code called hyperparams). If this composite score is greater than zero, the edge points towards the HomeTeam and vice versa, with the weight of this composite score (* see match_row_to_edge function). If it is zero, the edge is ignored. All these weighted directed edges are summed, such that all matches between two teams are represented by one weighted directed edge. A hardcoding with zeros at the entries for "FTG_diff" and "AvgOdds_diff" is performed to minimize risk of data leakage (* see random_search function). 

```python
# ── Data structure ───────────────────────────────────────────────────────────
class DirectedEdge:
    """A weighted directed edge from `source` to `target`."""
    def __init__(self, source: str, target: str, weight: float):
        self.source = source
        self.target = target
        self.weight = weight

# ── Edge construction ─────────────────────────────────────────────────────────

def match_stats(row) -> np.ndarray:
    """
    Extracts per-match differential statistics from a DataFrame row.
    Positive values favour the home team; the vector is L2-normalised.

    Parameters:
        row (pandas.Series): A row from the match DataFrame containing the
            following columns: FTHG, FTAG, HS, AS, HST, AST, HF, AF,
            HC, AC, HY, AY, HR, AR, AvgH, AvgA.

    Returns:
        np.ndarray: A normalised array of 8 differential statistics in the
            order [goals, shots, shots-on-target, fouls, corners,
            yellow-cards, red-cards, odds]. Returns the unnormalised array
            if the L2 norm is zero (e.g. all stats are equal).
    """
    stats = np.array([
        data['FTG_diff'][row.name],   # Goal difference
        data['SD'][row.name],      # Shot difference
        data['SOTD'][row.name],     # Shots on target difference
        data['F_diff'][row.name],      # Foul difference (away − home; fewer fouls is better)
        data['C_diff'][row.name],      # Corner difference
        data['YC_diff'][row.name],      # Yellow card difference (away − home; fewer cards is better)
        data['RC_diff'][row.name],      # Red card difference   (away − home; fewer cards is better)
        data['AvgOdds_diff'][row.name],    # Betting odds difference
    ])
    return stats
  
def match_row_to_edge(row, hyperparams) -> DirectedEdge | None:
    """
    Converts a match row into a directed edge pointing from loser → winner.
    Returns None for draws (zero weighted score).

    Parameters:
        row (pandas.Series): A row from the DataFrame
        hyperparams (np.ndarray): The hyperparameters for the weighted score calculation
    Returns:
        DirectedEdge: A directed edge from the losing team to the winning team with weight equal to 
            the absolute value of the weighted score, or None if it's a draw.
    """
    stats = match_stats(row)
    weighted_score = np.dot(stats, hyperparams)

    # *Direction of edge is based on sign of composite scores 
    if weighted_score > 0:
        return DirectedEdge(row['AwayTeam'], row['HomeTeam'],  weighted_score)
    elif weighted_score < 0:
        return DirectedEdge(row['HomeTeam'], row['AwayTeam'], - weighted_score) # minus sign to ensure weight is positive
    return None 

def compute_edges(data: pd.DataFrame, hyperparams) -> list[DirectedEdge]:
    """
    Applies `match_row_to_edge` to every row and drops draws.

    Parameters:
        data (pd.DataFrame): The DataFrame containing match data.
        hyperparams (np.ndarray): The hyperparameters for the weighted score calculation.
    Returns:
       list[DirectedEdge]: A list of directed edges representing the matches, excluding draws.
    
    """
    edges = data.apply(lambda row: match_row_to_edge(row, hyperparams), axis=1)
    return [e for e in edges if e is not None]

# ── Graph construction ────────────────────────────────────────────────────────

def edges_to_dataframe(edges: list[DirectedEdge]) -> pd.DataFrame:
    """
    Converts a list of DirectedEdge objects into an aggregated edge DataFrame.
    Multiple edges between the same pair of teams are summed.

    Parameters:
        edges (list[DirectedEdge]): A list of directed edges to convert.
    Returns:
        pd.DataFrame: A DataFrame with columns ['source', 'target', 'weight'] where weights are summed for duplicate edges.
    """
    df = pd.DataFrame({
        'source': [e.source for e in edges],
        'target': [e.target for e in edges],
        'weight': [e.weight for e in edges],
    })
    return df.groupby(['source', 'target'], as_index=False)['weight'].sum()

def build_graph(edges: list[DirectedEdge]) -> nx.DiGraph:
    """
    Builds a weighted directed graph from a list of DirectedEdge objects.
    
    Parameters:
        edges (list[DirectedEdge]): A list of directed edges to build the graph from.
    Returns:
        nx.DiGraph: A directed graph where nodes are teams and edges represent weighted interactions
    """
    edge_df = edges_to_dataframe(edges)
    return nx.from_pandas_edgelist(
        edge_df,
        source='source',
        target='target',
        edge_attr='weight',
        create_using=nx.DiGraph()
    )

# -─ Graph visualization ────────────────────────────────────────────────────────
def plot_graph(G: nx.DiGraph, title: str = 'Team Interaction Graph') -> None:
    """
    Plots a directed graph using a spring layout.
    
    Parameters:
        G (nx.DiGraph): The directed graph to plot.
        title (str): The title for the plot.
    """
    plt.figure(figsize=(10, 10))
    pos = nx.circular_layout(G)
    nx.draw(G, pos, with_labels=True, node_size=800,
            node_color='lightblue', edge_color='gray', font_size=7, alpha=0.8, width=0.5, arrowsize=10)
    plt.title(title)
    plt.show()  
```

```python
# ── Transition matrix and PageRank ─────────────────────────────────────────────
def build_transition_matrix(G: nx.DiGraph) -> np.ndarray:
    """
    Constructs the column-stochastic transition matrix from a directed graph.
    M = A^T · Δ⁻¹  (O8-LinkAnalysis slide 19)

    Parameters:
        G (nx.DiGraph): The directed graph representing team interactions.
    Returns:
        np.ndarray: The transition matrix
    """
    nodes = sorted(G.nodes())
    A     = nx.to_numpy_array(G, weight='weight', nodelist=nodes)
    Delta = np.diag(A.sum(axis=1))
    return A.T @ np.linalg.inv(Delta)


def power_iteration(
    M: np.ndarray,
    max_iter: int = 1000,
    alpha: float  = 0.85,
    tol: float    = 1e-6,
) -> np.ndarray:
    """
    Estimates the PageRank vector via power iteration with teleportation.
    (O8-LinkAnalysis slide 22)
    
    Parameters:
        M (np.ndarray): The column-stochastic transition matrix.
        max_iter (int): Maximum number of iterations to perform.
        alpha (float): Teleportation factor (probability of following a link vs. teleporting).
        tol (float): Convergence tolerance for the norm of rank changes.
    Returns:
        np.ndarray: A probability vector (sums to 1) over all nodes.
    """
    n   = M.shape[0]
    E   = np.ones((n, n)) / n
    M_  = alpha * M + (1 - alpha) * E          

    rank = np.ones(n) / n # Initial uniform distribution                 
    for _ in range(max_iter):
        new_rank = M_ @ rank
        if np.linalg.norm(new_rank - rank, 1) < tol:
            break
        rank = new_rank / new_rank.sum()

    return rank / rank.sum()


def compute_pagerank(G: nx.DiGraph, M: np.ndarray) -> list[tuple[str, float]]:
    """
    Computes the PageRank scores for each node in the graph.

    Parameters:
        G (nx.DiGraph): The directed graph representing team interactions.
        M (np.ndarray): The transition matrix derived from G.
    Returns:
        list[tuple[str, float]]: A list of (team, pagerank_score) tuples, sorted by score descending.

    """
    rank_vector = power_iteration(M)
    scores      = dict(zip(sorted(G.nodes()), rank_vector))
    return sorted(scores.items(), key=lambda x: x[1], reverse=True)


# ── Full PageRank pipeline ─────────────────────────────────────────────────────────────

def run_pagerank_pipeline(
    data: pd.DataFrame,
    hyperparams,
    plot: bool = False,
) -> tuple[nx.DiGraph, np.ndarray, list[tuple[str, float]]]:
    """
    Runs the full pipeline: data + hyperparams → edges → graph → transition matrix → PageRank ranking.

    Parameters:
        data (pd.DataFrame): The match data.
        hyperparams: The hyperparameters for edge weighting.
        plot (bool): Whether to plot the graph.

    Returns:
        G (nx.DiGraph): The constructed directed graph.
        M (np.ndarray): The transition matrix derived from G.
        sorted_teams (list[tuple[str, float]]): Teams sorted by PageRank score (descending).
    """
    edges = compute_edges(data, hyperparams)
    G = build_graph(edges)
    if plot:
        plot_graph(G)
    M = build_transition_matrix(G)
    sorted_teams = compute_pagerank(G, M)
    return G, M, sorted_teams

# ── Evaluation & search ───────────────────────────────────────────────────────

def evaluate(data: pd.DataFrame, hyperparams, norm_pts, loss: callable) -> float:
    """
    MSE between PageRank scores and normalised league points.
    
    Parameters:
        data (pd.DataFrame): The match data.
        hyperparams: The hyperparameters for edge weighting.
        norm_pts: List of (team, normalized_points) to compare against.
        loss: The loss function to evaluate the alignment between PageRank scores and normalized points.
    Returns:
        float: The computed loss value.
    """
    _, _, pagerank_scores = run_pagerank_pipeline(data, hyperparams)

    return loss(pagerank_scores, norm_pts)


def random_search(
    data: pd.DataFrame,
    norm_pts,
    n_trials:  int  = 500,
    data_leak: bool = False,
    loss: callable = score_loss
) -> tuple[np.ndarray, float]:
    """
    Randomly samples hyperparameter vectors and keeps the one with lowest MSE.
    Set data_leak=False to zero out goal-difference and odds weights.

    Parameters:
        data (pd.DataFrame): The match data.
        norm_pts: List of (team, normalized_points) to compare against.
        n_trials (int): Number of random hyperparameter vectors to sample.
        data_leak (bool): If False, zero out goal-difference and odds weights to prevent data leakage.
        loss (callable): The loss function to evaluate the alignment between PageRank
    Returns:
        tuple[np.ndarray, float]: The best hyperparameters found and their corresponding loss value.    

    """
    best_loss, best_params = float('inf'), np.ones(8)

    for _ in tqdm(range(n_trials), desc="Random Search"):
        params = np.random.uniform(0, 5, size=8)
        # *This is to prevent data leakage from the goal difference and betting odds feature
        if not data_leak:
            params[0] = 0   # Zero out goal-difference weight
            params[7] = 0   # Zero out betting-odds weight

        ls = evaluate(data, params, norm_pts, loss)
        if ls < best_loss:
            best_loss, best_params = ls, params.copy()
            
    return best_params, best_loss
```

In the following we will make a random search for the best hyperparameters to be used for the composite score. We use the unfinished 2025/2026 season as a validation set. **Evaluation of loss function and normalisation methods**:

The learned edge weights can be interpreted as measures of dominance in individual matchups, where stronger positive values indicate that a team statistically outperformed its opponent according to the selected in-game features.

Among the evaluated configurations, the score loss combined with total points normalisation produced the best overall performance for the sophisticatedgraph-based model. Compared to the baseline approach, the score loss decreased substantially, indicating that the weighted directed graph captures team strength more accurately than the simple unweighted graph. In particular, the score loss using total points normalisation improved from 0.000324 in the baseline model to 0.000048 in the sophisticated model.


The score loss was ultimately preferred over the rank loss because it provides a continuous measure of similarity between PageRank scores and the normalised league standings. In contrast, rank loss only measures differences in ordering and ignores the magnitude of those differences. As a result, score loss better reflects whether the graph-based representation captures the relative strength of teams across a season.
The choice of normalisation method also had a noticeable impact on the results. Points-per-match normalisation produced very small numerical differences between teams because all scores are bounded within a narrow interval (that is a small variance across teams). This makes the evaluation highly sensitive to small PageRank variations. In practice, this amplified the effect of teams with limited participation across seasons.


This issue, with the points-per-match normalisation, is particularly evident in the baseline bidirectional graph. Since every played match creates edges in both directions, teams with poor performance can still receive PageRank 'mass' simply by being connected to many opponents. For example, a team such as Blackpool F.C., which only appeared in a single Premier League season before relegation, remains reachable from all opposing teams despite losing many matches. Consequently, the random walk underlying PageRank can still frequently visit such teams, inflating their ranking relative to their actual performance. This ultimately explains why the loss increases for the sophisticated graph-based method when using points-per-match normalisation and score loss.


The directed weighted graph reduces this effect. Here, edges only point toward the statistically dominant team in a matchup, meaning that poorly performing teams receive far fewer incoming connections. In the case of Blackpool, only teams they genuinely outperformed contribute incoming PageRank mass. This produces rankings that align more closely with the observed league outcomes and reduces the bias introduced by weakly performing teams with sparse participation histories.

Overall, the combination of score loss and total points normalisation was selected as the primary evaluation framework. These measures provide a smoother and more informative comparison between PageRank scores and league performance, making them more suitable for our analysis objective.

## Team losses

The next two cells uses the earlier found best hyperparameters of all seasons i.e. the list **all_hyperparams**, to calculate the score loss of each team. This is done to reveal if some teams play in a way different from the other teams. 

This is done by running through all seasons and calculating each teams score loss in each season. Since the calculated team loss is summed over all the seasons a team has played the Premier League, each team loss is normalized by the number of seasons played by that team. 

```python
all_team_losses = []


# Fix global team universe (important for alignment)
all_teams = set()

# First pass to collect all teams across all seasons
for season in seasons:
    season_data = data[data['Season'] == season]
    norm_pts_season = normalize_points(
        compute_points_table(season_data),
        season_data
    )[0]
    all_teams.update([team for team, _ in norm_pts_season])

reference_teams = np.array(sorted(all_teams))

for hyp in range(len(all_hyperparams)):
    
    # Step 1: hyperparameters
    best_hyperparams = all_hyperparams[hyp]
    
    # Step 2: accumulate per-team loss across seasons
    team_loss_sum = {team: 0.0 for team in reference_teams}
    team_season_count = {team: 0 for team in reference_teams}
    
    for season in seasons:
        season_data = data[data['Season'] == season]
        
        norm_pts_season = normalize_points(
            compute_points_table(season_data),
            season_data
        )[0]
        norm_pts_dict = dict(norm_pts_season)
        
        pagerank = run_pagerank_pipeline(season_data, best_hyperparams)[2]
        pagerank_dict = dict(pagerank)
        
        for team in norm_pts_dict:
            if team not in pagerank_dict:
                continue
            
            pr_score = pagerank_dict[team]
            gt_score = norm_pts_dict[team]
            
            loss = (pr_score - gt_score) ** 2
            
            team_loss_sum[team] += loss
            team_season_count[team] += 1
    
    # Step 3: compute average loss per team (aligned vector)
    losses = np.zeros(len(reference_teams))
    
    for i, team in enumerate(reference_teams):
        if team_season_count[team] == 0:
            losses[i] = np.nan
        else:
            losses[i] = team_loss_sum[team] / team_season_count[team]
    
    all_team_losses.append(losses)

all_team_losses = np.array(all_team_losses)  # shape: (n_runs, n_teams)
```

## Spectral clustering on match statistics for comparison to part I clustering methods
To compare spectral clustering with the Part I methods (DBSCAN and K-means++) we take the 5-dimensional feature space where each match is represented as a point, and connect each match to its 9 nearest neighbors in feature space using the sklearn.neighbors module. This turns the match-statistics space into a sparse graph.

From this graph (represented by the adjacency matrix) we construct the degree matrix and the Laplacian matrix. We use the unnormalized Laplacian $L=D-A$ since it is the simplest formulation. We then compute the smallest eigenvalues and corresponding eigenvectors of the Laplacian using the eigsh function from scipy.sparse.linalg.

We use the eigenvectors associated with the smallest non-zero eigenvalues as an embedding and run K-means on that spectral representation.

This lets us compare whether the graph structure of the match statistics produces clearer clusters than DBSCAN and K-means++ in Part I.

```python
X = modified_stats.values # each match is a node, features come from Part I (modified_stats).
k = 9  # Number of nearest neighbors to connect. We use 9, since that was our MinPts for DBSCAN.

# 1) Creates adjacency matrix using nearest neigbors. (Is sparse, so only stores the non-zero entries --> the edges!)
A = kneighbors_graph(
    X,
    n_neighbors=k,
    mode="connectivity",   # 0/1 edges
    include_self=False,
    metric="euclidean"     # To use the same distance as in part I.
)

# 2) Makes the graph undirected.
A_undirected = A.maximum(A.T)

# 3) Computes the degree matrix
A = A_undirected.tocsr()
deg = np.asarray(A.sum(axis=1)).ravel()
D = sp.diags(deg, offsets=0, format="csr")

# 4) computes the laplacian matrix
L = D - A
```

```python
# 5) --- We compute the smallest eigenvalues of L ---
m = 4 ## Number of eigenvalues to compute using eigsh. We want the 3 smallest eigenvalues for clustering into 3 groups. 3 + 1 (since the first is usually 0/everything).

# eigsh solves for the eigenvalues of the L matrix. It computes the m eigenvalues closest to sigma (0.0) --> the 10 smallest eigenvalues.
evals, evecs = sp.linalg.eigsh(L, k=m, sigma=0.0, which="LM")

# Sorts the eigenvalues in order.
order = np.argsort(evals)
evals = evals[order]
evecs = evecs[:, order]

print("Smallest eigenvalues of L:")
print(evals)
```

```python
# --- does the spectral clustering using the eigenvectors ---  Does kmeans on the eigenvectors to cluster the data into k clusters. 

k_clusters = 3

# Skips the first eigenvector (corresponding to the smallest eigenvalue --> zero) and uses the next k eigenvectors as features for k-means.
U = evecs[:, 1 : k_clusters + 1] #since the first eigenvector is usually the trivial one (all ones) corresponding to the zero eigenvalue.

# Settings for k-means --> 3 clusters, fixed random seed for reproducibility.
kmeans_spec = KMeans(n_clusters=k_clusters, random_state=RANDOM_SEED, n_init="auto")

# Do k-means on the rows of U to get cluster labels.
labels_spec = kmeans_spec.fit_predict(U)

print("Spectral Clustering labels:", np.unique(labels_spec))

# Find contingency table and purity.
c_table = contingency_table(labels_spec, ground_truth, ignore_noise=False)
purity = purity_score(c_table)

print("Contingency table:\n", c_table)
print("Purity per cluster:", purity)

# Print ARI and NMI.
clustering_result3 = {
    'Spectral Graph Clustering': labels_spec,
}
for name, y_pred in clustering_result3.items():
    print(f"\n{name}:")
    ari = adjusted_rand_score(ground_truth, y_pred)
    nmi = normalized_mutual_info_score(ground_truth, y_pred)
    print(f"ARI: {ari:.4f}, NMI: {nmi:.4f}")
```

# Pattern mining (Part III)

**Pattern mining objective:**

The objective of the pattern mining analysis is to explore whether association rule mining can uncover meaningful relationships between bookmakers' odds across different seasons. By discretizing the odds features into categorical bins, we aim to identify frequent itemsets and generate association rules that reveal how certain odds patterns co-occur across matches. This analysis can provide insights into the consistency of bookmakers' odds and potentially highlight any systematic biases or trends in their offerings over time.

## Data loading and preprocessing and feature engineering for pattern mining
The idea is to do a discretization of the odds features to see if we can mine association rules between the bookmakers. Note that there is a slight difference in the odds features across the seasons, as some seasons have odds from more bookmakers than others. To address this we choose a subset of seasons where the same bookmakers are represented.

```python
def load_common_data(folder_path, min_files=10, encoding="latin1", print_progress=False):
    """
    Loads and concatenates CSV files that share a common set of columns.

    The function scans all CSV files in the specified folder, identifies their
    available columns, and greedily selects files such that the intersection of
    shared columns is maximized while including at least `min_files` files.
    Only the common columns are loaded and concatenated into a single dataframe.

    Parameters:
        folder_path (str): Path to the directory containing the CSV files.
        min_files (int): Minimum number of compatible CSV files to include in the final dataset.
        encoding (str): File encoding used when reading the CSV files. Default is "latin1".
        print_progress (bool): If True, prints information during execution.

    Returns:
        final_df (pd.DataFrame): Concatenated dataframe containing only the shared columns across the
            selected files, with an additional "Season" column indicating the source file.
        selected_files (list[str]): List of filenames included in the final merged dataset.
        current_common (list[str]): List of column names shared across all selected files.
    """
    file_columns = {}

    # --- Step 1: Read headers safely ---
    for file in os.listdir(folder_path):
        if file.endswith(".csv"):
            path = os.path.join(folder_path, file)
            try:
                df = pd.read_csv(path, nrows=0, encoding=encoding)
                file_columns[file] = set(df.columns)
            except Exception as e:
                print(f"Skipping {file} during scan: {e}")

    if len(file_columns) < min_files:
        raise ValueError(f"Not enough readable files. Found {len(file_columns)}")

    # --- Step 2: Start with file that has the most columns ---
    sorted_files = sorted(file_columns.items(), key=lambda x: len(x[1]), reverse=True)

    selected_files = [sorted_files[0][0]]
    current_common = set(sorted_files[0][1])

    if print_progress:
        print(f"\n Starting with: {os.path.splitext(selected_files[0])[0]} ({len(current_common)} cols)")

    remaining_files = {f: cols for f, cols in sorted_files[1:]}

    # --- Step 3: Greedily add files that maximize remaining common columns ---
    while len(selected_files) < min_files and remaining_files:
        best_file = None
        best_intersection = set()

        for f, cols in remaining_files.items():
            intersection = current_common.intersection(cols)

            if len(intersection) > len(best_intersection):
                best_file = f
                best_intersection = intersection

        if best_file is None or len(best_intersection) == 0:
            print("No more compatible files can be added without losing all columns.")
            break

        selected_files.append(best_file)
        current_common = best_intersection

        if print_progress:
            print(f" + Added: {os.path.splitext(best_file)[0]} → {len(current_common)} common cols")

        del remaining_files[best_file]

    # --- Final report ---
    if print_progress:
        print(f"\n Final selected {len(selected_files)} files:")
    for f in selected_files:
        if print_progress:
            print(f" - {os.path.splitext(f)[0]} ({len(file_columns[f])} cols)")

    if print_progress:
        print(f"\n Final shared columns: {len(current_common)}")
        print(sorted(current_common))

    # --- Step 4: Load data ---
    dataframes = []

    for file in selected_files:
        path = os.path.join(folder_path, file)
        try:
            df = pd.read_csv(path, usecols=lambda c: c in current_common, encoding=encoding)

            season = os.path.splitext(file)[0]
            df["Season"] = season

            dataframes.append(df)
        except Exception as e:
            print(f"Skipping {file} during load: {e}")

    final_df = pd.concat(dataframes, ignore_index=True)

    return final_df, selected_files, list(current_common)

final_df, num_cols = [], []
for i in range(1, 24):
    df, _, cols = load_common_data(dataPath, min_files=i)
    final_df.append(df)
    num_cols.append(len(cols))    
plt.plot(range(1, 24), num_cols, marker='o')
plt.xlabel('Number of Files Included')
plt.ylabel('Number of Common Columns')
plt.title('Common Columns vs. Number of Files Included')
plt.grid()
plt.show()

```

The process starts with the season containing the most columns, then iteratively adds seasons that preserve the largest possible overlap of shared columns. At each step, the set of common columns is updated to ensure all selected seasons remain compatible.

After selecting the final group of seasons, only the columns shared across all of them are kept, and each season is loaded into a single combined DataFrame.

This ensures a consistent feature set across seasons while maximizing the number of usable datasets included.

### Feature engineering and discretization
In order to do pattern mining on the odds features, we need to discretize them. Therefore we needed to investigate the distribution of the odds features to find a good way to discretize them.

Based on the distributions above, it does not make sense to use a single discretization scheme for all columns, as *draw* odds differ substantially from *home* and *away* odds. We therefore construct separate bins per outcome type (`H`, `D`, `A`).

Instead of binning each column independently, we pool odds across all bookmakers for each outcome and compute quantile-based bins. This ensures consistent bins across bookmakers while capturing the distinct distributional shape of each outcome type.

```python
def compute_quantile_bins(data, relevant_cols, suffixes=["H", "D", "A"], n_bins=10):
    """
    Compute quantile-based bin edges for each outcome type (H, D, A),
    pooling odds across all bookmakers for that outcome type.
    Parameters:
        data (pd.DataFrame): DataFrame containing the odds columns
        relevant_cols (list): List of column names to consider for pooling
        suffixes (list): List of outcome suffixes
        n_bins (int): Number of quantile bins to create (default 5)

    Returns:
        A dict like: {'H': (bins, labels), 'D': (bins, labels), 'A': (bins, labels)}
    """
    bin_specs = {}
    
    for suffix in suffixes:
        # Pool all odds for this outcome across all bookmakers
        cols_for_suffix = [c for c in relevant_cols if c.endswith(suffix)]
        pooled = pd.concat([data[c].dropna() for c in cols_for_suffix])
        
        # Compute quantile edges (0%, 20%, 40%, 60%, 80%, 100%)
        quantiles = np.linspace(0, 1, n_bins + 1)
        edges = pooled.quantile(quantiles).values
        
        # Ensure strictly increasing edges (can have ties at extremes)
        edges = np.unique(edges)
        edges[0] = 0       # open lower bound
        edges[-1] = np.inf # open upper bound
        
        # Labels like "1.42–1.75", "1.75–2.10", etc.
        labels = [
            f"{edges[i]:.2f}–{edges[i+1]:.2f}".replace("inf", "∞")
            for i in range(len(edges) - 1)
        ]
        
        bin_specs[suffix] = (edges, labels)
        print(f"[{suffix}] edges: {np.round(edges, 3)}  →  {len(labels)} bins")
    
    return bin_specs


def discretize_odds_quantile(series, bin_specs):
    """
    Discretize a single odds column using precomputed quantile bin specs.
    Detects outcome type from the column name's last character.
    Parameters:
        series (pd.Series): pandas Series of odds to discretize
        bin_specs (dict): dict of {suffix: (edges, labels)} for discretization
    Returns:
        pd.Series: pandas Series of categorical bin labels corresponding to the input odds
    """
    suffix = series.name[-1]  # 'H', 'D', or 'A'
    bins, labels = bin_specs[suffix]
    return pd.cut(series, bins=bins, labels=labels)

# Compute shared bins per outcome type, pooled across bookmakers
bin_specs = compute_quantile_bins(data, relevant_cols, n_bins=10)

# Apply to each column
discrete_data = data.copy()
for c in relevant_cols:
    if c[-1] in bin_specs:
        discrete_data[c] = discretize_odds_quantile(discrete_data[c], bin_specs)
```

### Applying the the Apriori algorithm
The Apriori algorithm is applied to the discretized odds data to identify frequent itemsets and generate association rules. The algorithm iteratively builds itemsets of increasing size, starting from single items (individual odds categories) and expanding to larger combinations. At each step, it prunes itemsets that do not meet a specified minimum support threshold. Once frequent itemsets are identified, association rules are generated. Here we defined a function *is_relevant* to filter the generated rules based on certain criteria to filter out irrelevant or trivial rules.

```python
def make_transactions(df, cols):
    """
    Converts the discretized odds columns into a list of transactions for association rule mining.
    Each transaction is a list of items like "B365H=0-1.34" representing the bin for that column in that row.

    Parameters:
        df (pd.DataFrame): The dataframe containing the discretized odds columns.
        cols (list): List of column names to include in the transactions.
    Returns:
        list of lists: Each inner list is a transaction containing items
    """
    T = []

    for _, row in df[cols].iterrows():
        items = []

        for c in cols:
            val = row[c]
            if pd.notna(val):
                items.append(f"{c}={val}")

        T.append(items)

    return T
T = make_transactions(discrete_data, relevant_cols)

rule  = lambda lhs, rhs: "%s => %s" % (str(lhs), str(rhs)) # For generating rule strings

def compute_support(Ck: list, T: list, still_applicable=None):
    """
    Function that computes the support of each candidate itemset in Ck with respect to the transactions in T.
    Parameters:
        Ck (list): list of candidate itemsets of size k
        T (list): list of transactions (each transaction is a list of items)
        still_applicable (list): a list of boolean values indicating whether each transaction is applicable
    Returns:
        counts (dict): a dictionary of counts corresponding to the support of each candidate itemset in Ck
    """
    if still_applicable is None:
        still_applicable = [True] * len(T)
        
    counts = {} # Dictionary to store counts of each candidate itemset slide 40
    for row, t in enumerate(T):
        if not still_applicable[row]: continue
        found_any = False
        for c in Ck:
            if set.issubset(set(c), t):
                if c in counts:
                    counts[c] += 1
                else:
                    counts[c] = 1
                found_any = True
        still_applicable[row] = found_any
    return counts

def extend_prefix_tree(Ck_prev):
    """
    Function that extends the prefix tree by generating new candidate itemsets of size k+1 from the previous candidate itemsets of size k.
    Parameters:
        Ck_prev (list): list of candidate itemsets of size k
    Returns:
        Ck (list): list of candidate itemsets of size k+1
    """
    Ck = set() 
    #join step
    for itemset in Ck_prev:
        its1 = tuple(sorted(itemset))
        for itemset2 in Ck_prev:
            its2 = tuple(sorted(itemset2))
            if its1[:-1] == its2[:-1] and its1 < its2: Ck.add(its1 + its2[-1:]) # Join step: if the first k-1 items are the same and the last item is different, we can join them
    #prune step
    to_remove = set()
    for c in Ck:
        for subset in combinations(c, len(c) - 1):
            if subset not in Ck_prev:
                to_remove.add(c)
                break            
    for c in to_remove:
        Ck.remove(c)
    return Ck

def powerset(iterable):
    """
    Function that generates the powerset of a given iterable.
    Parameters: 
            iterable (iterable (e.g., list, set, etc.)): for which to generate the powerset
    Returns:
            A list of all subsets of the input iterable      
    """
    s = list(iterable)
    out = chain.from_iterable(combinations(s, r) for r in range(1, len(s))) # Generate all non-empty subsets
    return out

def is_relevant(A, B):
    def parse(item):
        col, val = item.split("=")
        bookmaker = col[:-1]
        outcome = col[-1]
        value = val.replace(" ", "")
        return bookmaker, outcome, value

    A_parsed = [parse(x) for x in A]
    B_parsed = [parse(x) for x in B]

    all_parsed = A_parsed + B_parsed

    # 1. Reject if rule involves all three outcomes (A, H, D)
    outcomes = {o for _, o, _ in all_parsed}
    if outcomes == {"A", "H", "D"}:
        return False
    
    # 2. Remove same-market duplication across bookmakers
    A_market = {(o, v) for _, o, v in A_parsed}
    B_market = {(o, v) for _, o, v in B_parsed}
    if A_market & B_market:
        return False

    # 3. Reject “mirrored” home/away rules
       #If any outcome in A is 'H' and in B is 'A' (or vice versa), reject
    A_outcomes = {o for _, o, _ in A_parsed}
    B_outcomes = {o for _, o, _ in B_parsed}
    if ("H" in A_outcomes and "A" in B_outcomes) or ("A" in A_outcomes and "H" in B_outcomes):
        return False
    
    return True

def apriori_algorithm(T, support=0.05, min_confidence=0.7):
    """
        Apriori algorithm for mining association rules.
        Parameters:
            T:               A list of lists, each inner list will contiain integer-item-ids. 
                             Example: T = [[1, 2, 5], [2, 3, 4], [1, 6]]
            support:         The proportion of occurences needed to keep itemsets.
            min_confidence:  Minimum confidence for the algorithm to output the rule.
        Returns:
            rules ([(str, float)]): A list of tuples (rule_str, confidence) 
    """        
    rules = []
    # --- Generate frequent itemsets using the Apriori algorithm ---
    N = len(T) # Total number of transactions
    # Construct the initial candidate itemsets of size 1
    items = set(i for t in T for i in t) # Get all unique items in the transactions
    C1 = set((i,) for i in items) # Convert to set of tuples for consistency with later itemsets
    counts = compute_support(C1, T) # Compute support for itemsets of size 1
    L1 = [c for c in counts if counts[c] / N >= support] # Filter itemsets by support
    L = [L1]
    support_dict = {c: counts[c] / N for c in L1} # Store support values for itemsets of size 1
    # Second step
    k = 2
    while L[k-2]: # While there are itemsets of size k-1
        Ck = extend_prefix_tree(L[k-2]) # Generate candidate itemsets of size k
        if not Ck: break # If no candidates, stop

        counts = compute_support(Ck, T) # Compute support for itemsets of size k
        Lk = []
        for i in Ck:
            supp = counts.get(i, 0) / N # Get support for itemset i, default to 0 if not found
            if supp >= support:
                Lk.append(i) # Keep itemsets that meet the support threshold
                support_dict[i] = supp # Store support value for itemset i
        if not Lk: break # If no itemsets meet the support threshold, stop
        L.append(Lk) # Add the frequent itemsets of size k to the list
        k += 1    
    # --- Generate rules from the frequent itemsets ---  
    for Lk in L:
        for itemset in Lk:
            if len(itemset) < 2: continue # We need at least 2 items to form a rule

            for A in powerset(itemset):
                B = tuple(sorted(set(itemset) - set(A))) # Get the remaining items as B
                if not B: continue # Skip if B is empty

                if not is_relevant(A, B):
                    continue

                conf = support_dict[itemset] / support_dict[A] # Calculate confidence of the rule A => B
                if conf >= min_confidence:
                    rules.append((rule(A, B), conf)) # Store the rule along with its confidence
    return rules

def pretty_print_rules(rules):
    """
    Pretty-print association rules as a table.
    Parameters:
        rules ([(str, float)]): list of tuples (rule_str, confidence)
    Returns:
        None (displays a sorted DataFrame of rules)    
    """
    data = []
    for r, conf in rules:
        # Parse rule string of the form "('A',) => ('B',)"
        lhs, rhs = r.split(" => ")
        lhs = lhs.strip("()").replace("'", "").replace(",", "")
        rhs = rhs.strip("()").replace("'", "").replace(",", "")
        data.append({"LHS": lhs, "RHS": rhs, "Confidence": round(conf, 3)})

    df = pd.DataFrame(data)
    display(df.sort_values("Confidence", ascending=False).reset_index(drop=True))

support = 0.08
min_confidence = 0.70                
rules = apriori_algorithm(T, support=support, min_confidence=min_confidence)
print(f"Generated {len(rules)} rules with support >= {support} and confidence >= {min_confidence}")
print("Sample rules:")
pretty_print_rules(rules)
```

## Evaluation of the generated rules
Generally it is hard to find a good definition of a non trivial rule in this context. We made a function that filters out trivial rules based on the following criteria:
- **Reject rules involving all three outcomes (A, H, D):**  
  If a rule includes "Away win (A)", "Home win (H)", and "Draw (D)" simultaneously, it is rejected.
  This avoids rules like: 

  **('B365A=8.75–∞', 'B365H=0.00–1.34') => ('PSD=5.77–∞',)** 

  This rule is trivial because it simply states that if the odds for an away win are very high and the odds for a home win are very low, then the odds for a draw are also very high, which is not insightful.


- **Remove duplication across bookmakers:**  
  If any `(outcome, value)` pair appears in both side A and side B, the rule is rejected. This rule prevents trivial rules that simply restate the same condition across different bookmakers, such as: 

  **(('BWA=8.75–∞',) => ('PSA=8.75–∞',), 0.98)**. 
  
  This rule is trivial because it just says that if one bookmaker has high odds for an away win, then another bookmaker also has high odds for an away win, which is not insightful.


- **Reject home/away rules:**  
  If side A contains "H" and side B contains "A", or vice versa, the rule is rejected. This prevents trivial rules that simply mirror home and away conditions, such as: 
  
  **(('BWH=5.25–∞',) => ('PSA=0.00–1.63',), 0.96)**. 
  
  This rule is trivial because it just states that if the odds for a home win are very high, then the odds for an away win are very low, which is not insightful.

With these rejections in place and by tuning the parameters support and min_confidence, we succeeded in filtering out enough rules to be able to manually inspect the remaining rules for interesting patterns. We found some interesting patterns, such as:

**(('PSD=5.77–∞',) => ('PSA=8.75–∞',), 0.75)**

This rule suggests that when the draw is considered very unlikely by Pinnacle, the odds for an away win are very high (75% of the time). By extension, this implies that a home win becomes more likely under the same conditions. This pattern indicates that Pinnacle may systematically favor home teams in its odds. This pattern can also be found directly on the form:

**(('PSD=5.77–∞',) => ('PSH=0.00–1.34',), 0.78)**

and with different bookmakers involved:

**(('PSD=5.77–∞',) => ('BWH=0.00–1.34', 'PSH=0.00–1.34'), 0.724)**
